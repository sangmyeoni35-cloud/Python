```python
class Account:
    def __init__(self):
        # 변수 이름 앞에 __를 붙이면 외부에서 직접 수정하지 못하게 보호(캡슐화)됩니다.
        self.__balance = 0 

    def deposit(self, amount):
        self.__balance += amount

    def withdraw(self, amount): # 오타 수정: withrow -> withdraw
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

# 1. 인스턴스 생성 (반드시 괄호 () 필요)
a2 = Account() 

# 2. 메서드 호출
a2.deposit(50)

# 3. 데이터 출력
# a2.__balance 라고 하면 에러가 납니다. 약속된 통로(메서드)를 이용하세요.
print(a2.get_balance())
```
